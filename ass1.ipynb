```python
password = input("Enter your password: ")

has_digit = False
has_uppercase = False

for char in password:
    if char.isdigit():
        has_digit = True
    if char.isupper():
        has_uppercase = True

if len(password) >= 8 and has_digit and has_uppercase:
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
sentence = input("Enter a sentence: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0

for char in sentence:
    if char.lower() in "aeiou":
        vowels += 1
    elif char.isalpha():
        consonants += 1
    elif char.isdigit():
        digits += 1
    elif char == " ":
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```

```python
grades = [75, 48, 90, 62, 35, 88, 51]

passed = 0
failed = 0

for grade in grades:
    print(grade)

    if grade >= 50:
        passed += 1
    else:
        failed += 1

print("Passed students:", passed)
print("Failed students:", failed)
```

```python
numbers = [15, 42, 8, 91, 23, 4, 55]

largest = numbers[0]
smallest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

    if number < smallest:
        smallest = number

print("Largest number:", largest)
print("Smallest number:", smallest)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```

```python
numbers = [5, 2, 5, 8, 2, 10, 8, 12]

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

print(unique_numbers)
```

```python
words = ["apple", "banana", "avocado", "blueberry", "cherry", "apricot"]

grouped_words = {}

for word in words:
    first_letter = word[0]

    if first_letter in grouped_words:
        grouped_words[first_letter].append(word)
    else:
        grouped_words[first_letter] = [word]

print(grouped_words)
```

```python
cart = {
    "Laptop": 25000,
    "Mouse": 500,
    "Keyboard": 1200,
    "Headphones": 900
}

total_price = 0
expensive_products = 0

for product, price in cart.items():
    print(product, ":", price)

    total_price += price

    if price > 1000:
        expensive_products += 1

print("Total price:", total_price)
print("Products above 1000:", expensive_products)

if total_price > 25000:
    print("Expensive Order")
else:
    print("Normal Order")
```

```python
students = {
    "Ahmed": [70, 80, 90],
    "Sara": [95, 88, 92],
    "Omar": [40, 55, 45],
    "Mona": [60, 75, 70]
}

passed_students = []
failed_students = []

for name, grades in students.items():
    average = sum(grades) / len(grades)

    print("Student:", name)
    print("Average:", average)

    if average >= 50:
        print("Passed")
        passed_students.append(name)
    else:
        print("Failed")
        failed_students.append(name)

    print("----------------")

print("Passed students:", passed_students)
print("Failed students:", failed_students)
```
